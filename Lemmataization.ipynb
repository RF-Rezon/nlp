## WordNet Lemmatizer

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
words = ['eating','eats','eaten','writing','writes', 'programming','doing']
```

```python
lemma = WordNetLemmatizer()
```

```python
for word in words:
    print(word+"------->"+lemma.lemmatize(word, pos='v'))
```
