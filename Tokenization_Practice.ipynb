```python
corpus = """Hello all! Rejwan here.
Take love from me. 
I want to be a ai engineer. Here's all my strategies."""
```

```python
print(corpus)
```

## Paragaraph --> Sentence 

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize,TreebankWordTokenizer
```

```python
documents  = sent_tokenize(corpus)
```

```python
for single_sentence in documents:
    print(single_sentence)
```

## Sentence --> Words

```python
for single_sentence in documents:
    print(word_tokenize(single_sentence))
```

```python
for single_sentence in documents:
    print(wordpunct_tokenize(single_sentence))
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
