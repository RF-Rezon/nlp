## Stemming

```python
words = ['eating','eats','eaten','writing','writes', 'programming','programes','programmer']
```

## PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer(mode='NLTK_EXTENSIONS')
```

```python
for word in words:
    print(word+"------>"+stemmer.stem(word, to_lowercase=True))
```

## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowball = SnowballStemmer('english', ignore_stopwords=False)
```

```python
for word in words:
    print(word+"------>"+snowball.stem(word))
```

## RegexStemmer

```python
from nltk.stem import RegexpStemmer
```

```python
regexS = RegexpStemmer('ing$|en$|s$|es$|er$', min=0)
```

```python
for word in words:
    print(word+"------>"+regexS.stem(word))
```
